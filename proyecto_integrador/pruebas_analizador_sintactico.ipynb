```python
%%bash
set -e

carpeta=$(mktemp -d)
git clone --depth 1 --branch analizador-Leslie \
  https://github.com/LeslieTamara14/tecnm-lenguajes-automatas-2.git "$carpeta"

cd "$carpeta/proyecto_integrador"
make

printf 'int x = 1 + 2 * 3;\n' > prueba.lj
./lenguajito prueba.lj --ast

# Caso inválido: falta el punto y coma.
printf 'int x = 1 + 2 * 3\n' > prueba_invalida.lj
estado=0
diagnostico=$(./lenguajito prueba_invalida.lj --ast) || estado=$?
printf '%s\n' "$diagnostico"
test "$estado" -eq 1
grep -q 'error S01:' <<< "$diagnostico"
```
